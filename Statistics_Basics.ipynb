```python
#Question 1: What is the difference between descriptive statistics and inferential
#statistics? Explain with examples.
```

- 1. Descriptive Statistics: It summarizes and presents the data we have using the mean, median, mode, tables, and graphs.
Example: Calculating the average marks of 30 students in a class.
2. Inferential Statistics: It uses data from a sample to make predictions or draw conclusions about a larger population.
Example: Surveying 100 voters to predict the election results of an entire city.

```python
#Question 2: What is sampling in statistics? Explain the differences between random
#and stratified sampling.
```

- Sampling is the process of selecting a small group of individuals from a larger population to study and draw conclusions.
1. Random Sampling: Every individual in the population has an equal chance of being selected.
Example: Randomly selecting 50 students from a school.
2. Stratified Sampling: The population is divided into groups (strata) based on characteristics, and samples are selected from each group.
Example: Dividing students by class (Class 8, 9, and 10) and selecting students from each class.

```python
#Question 3: Define mean, median, and mode. Explain why these measures of central
#tendency are important.
```

- 1. Mean: The sum of all values divided by the total number of values.
Example: 2, 4, 6 → Mean = (2 + 4 + 6) / 3 = 4.
2. Median: The middle value when data is arranged in ascending or descending order.
Example: 2, 4, 6 → Median = 4.
3. Mode: The value that occurs most frequently in a dataset.
Example: 2, 4, 4, 6 → Mode = 4.

```python
#Question 4: Explain skewness and kurtosis. What does a positive skew imply about the
#data?
```

- 1. Skewness: Skewness measures the asymmetry of a data distribution. It can be positive, negative, or zero.
2. Kurtosis: Kurtosis measures the heaviness of a distribution’s tails and its tendency to produce extreme values compared with a normal distribution.
Positive Skew: A positive skew means the distribution has a long tail on the right side. Most values are concentrated on the left, with a few unusually high values.

```python
#Question 5: Implement a Python program to compute the mean, median, and mode of
#a givenlist of numbers.
#numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]
#Include your Python code and output in the code box below.)
```

```python
import statistics

numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]

mean = statistics.mean(numbers)
median = statistics.median(numbers)
mode = statistics.multimode(numbers)

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
```

```python
#Question 6: Compute the covariance and correlation coefficient between the following
#two datasets provided as lists in Python:
#list_x = [10, 20, 30, 40, 50]
#list_y = [15, 25, 35, 45, 60]
#(Include your Python code and output in the code box below.)
```

```python
import numpy as np

list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

covariance = np.cov(list_x, list_y)[0, 1]
correlation = np.corrcoef(list_x, list_y)[0, 1]

print("Covariance:", covariance)
print("Correlation coefficient:", correlation)
```

```python
#Question 7: Write a Python script to draw a boxplot for the following numeric list and
#identify its outliers. Explain the result:
#data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]
#(Include your Python code and output in the code box below.)
```

```python
import matplotlib.pyplot as plt

data = [12, 14, 14, 15, 18, 19, 19, 21,
        22, 22, 23, 23, 24, 26, 29, 35]

plt.boxplot(data)
plt.title("Boxplot and Outlier Detection")
plt.ylabel("Values")
plt.show()

# Identify outliers using the IQR method
import numpy as np

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = [x for x in data if x < lower or x > upper]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower)
print("Upper bound:", upper)
print("Outliers:", outliers)
```

```python
#Question 8: You are working as a data analyst in an e-commerce company. The
#marketing team wants to know if there is a relationship between advertising spend and
#daily sales.
#● Explain how you would use covariance and correlation to explore this
#relationship.
#● Write Python code to compute the correlation between the two lists:
#advertising_spend = [200, 250, 300, 400, 500]
#daily_sales = [2200, 2450, 2750, 3200, 4000]
#(Include your Python code and output in the code box below.)
```

```python
import numpy as np

advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]

correlation = np.corrcoef(advertising_spend, daily_sales)[0, 1]

print("Correlation coefficient:", correlation)
```
